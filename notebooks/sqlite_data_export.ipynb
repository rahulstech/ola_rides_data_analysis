```python
import sqlite3
import pandas as pd 
import os

data_path = "../data/OLA_Rides_Clean.xlsx"

sql_path = "../sql/ola_rides.sqlite3"

df = pd.read_excel(data_path)

# modify column name before export to databse
df_sql = df.rename(columns={
    'Vehicle Images': 'Vehicle_Images'
})

df_sql.columns = df_sql.columns.str.lower()

# create sqlite connection
with sqlite3.connect(sql_path) as conn:

    # export DataFrame to database
    df_sql.to_sql(
        name='rides',
        con=conn,
        if_exists='replace',
        index=False # True = numeric column index also added
    )

    cursor = conn.cursor()

    cursor.execute("SELECT COUNT(`booking_id`) AS `row_count` FROM `rides`;")

    # check DataFrame and database row counts to ensure successful export
    print(f"Total rows found in {data_path}: {len(df)}")
    print(f"Total rows found in {sql_path}: {cursor.fetchone()[0]}")

```
