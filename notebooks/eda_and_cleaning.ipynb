```python
import pandas as pd

data_root = '../data'

df = pd.read_excel(f"{data_root}/OLA_DataSet.xlsx")

df.info()
```

```python
export_columns = [
    'Booking_ID','Date','Customer_ID','Booking_Status',
    'Canceled_Rides_by_Customer', 'Canceled_Rides_by_Driver',
    'Incomplete_Rides', 'Incomplete_Rides_Reason', 'Booking_Value', 
    'Ride_Distance', 'Payment_Method', 'Customer_Rating', 'Driver_Ratings',
    'Vehicle_Type', 'Vehicle Images'
]

df_clean = df[export_columns]

df_clean.to_excel(
    f"{data_root}/OLA_Rides_Clean.xlsx",
    engine="openpyxl",
    sheet_name="rides",
    index=False
)
```
